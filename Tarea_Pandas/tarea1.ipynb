```python
import pandas as pd

# 1.A Carga el archivo student-mat.csv en un DataFrame
df = pd.read_csv('student-mat.csv')

# 1.B Crea una columna llamada student_id utilizando el índice original del DataFrame
df['student_id'] = df.index

# Mostrar que se cargó correctamente viendo las primeras 5 filas
df.head()
```

```python
# 1.C Revisa las dimensiones de la tabla (número de filas y columnas)[span_2](start_span)[span_2](end_span)
print("Dimensiones de la tabla (filas, columnas):")
print(df.shape)
print("-" * 30)

# 1.D Identifica el tipo de datos de cada columna[span_3](start_span)[span_3](end_span)
print("\nTipos de datos de cada columna:")
print(df.dtypes)
print("-" * 30)

# 1.F Obtén estadísticas descriptivas de las columnas numéricas[span_4](start_span)[span_4](end_span)
# (Nota: El inciso 1.E ya lo hiciste arriba con df.head(), solo que te mostró 5 filas por defecto. 
# Si quieres ver 10, puedes cambiarlo a df.head(10) en la celda de arriba).
print("\nEstadísticas descriptivas:")
df.describe()
```

```python
import numpy as np

# 2.A Seleccionar aleatoriamente el 5% de las posiciones numéricas y ponerlas nulas (NaN)[span_2](start_span)[span_2](end_span)
# Primero identificamos cuáles son las columnas numéricas
columnas_numericas = df.select_dtypes(include=['number']).columns

# Calculamos cuántas filas representan ese 5%
n_nulos = int(len(df) * 0.05)

# Por cada columna numérica, elegimos filas al azar y las volvemos nulas (NaN)
for col in columnas_numericas:
    indices_aleatorios = np.random.choice(df.index, size=n_nulos, replace=False)
    df.loc[indices_aleatorios, col] = np.nan

print("✅ 2.A - Valores nulos (NaN) insertados con éxito.\n")


# 2.B Cuenta la cantidad de valores nulos por columna y por fila[span_3](start_span)[span_3](end_span)
print("--- 2.B Cantidad de nulos por columna ---")
print(df.isna().sum())

print("\n--- 2.B Cantidad de nulos por fila (mostrando las primeras 10) ---")
print(df.isna().sum(axis=1).head(10))


# 2.C Aplicar reglas de limpieza[span_4](start_span)[span_4](end_span)
# a. Eliminar filas si hay nulos en 'sex', 'school', 'address[span_5](start_span)'[span_5](end_span)
df = df.dropna(subset=['sex', 'school', 'address'])

# b. Reemplazar nulos en numéricas con la mediana de la columna[span_6](start_span)[span_6](end_span)
# (Aplicamos la limpieza a todas las numéricas que ensuciamos en el paso A para repararlo todo)
for col in columnas_numericas:
    mediana = df[col].median()
    df[col] = df[col].fillna(mediana)

print("\n✅ 2.C - Reglas de limpieza aplicadas (filas borradas y medianas calculadas).\n")


# 2.D Verifica que al final ya no existan valores nulos[span_7](start_span)[span_7](end_span)
total_nulos_restantes = df.isna().sum().sum()
print(f"--- 2.D Verificación ---")
print(f"Total de nulos al finalizar la limpieza: {total_nulos_restantes}")

if total_nulos_restantes == 0:
    print("¡Felicidades! Tu DataFrame está completamente limpio otra vez.")
```

```python
# --- 3. Filtrado y selección ---

# 3.A Crea una columna llamada avg_grade: (G1+G2+G3)/3[span_2](start_span)[span_2](end_span)
df['avg_grade'] = (df['G1'] + df['G2'] + df['G3']) / 3

# 3.B Selecciona a los estudiantes con avg_grade > 12[span_3](start_span)[span_3](end_span)
estudiantes_promedio_alto = df[df['avg_grade'] > 12]
print(f"Estudiantes con promedio mayor a 12: {len(estudiantes_promedio_alto)}")

# 3.C Selecciona a los estudiantes con consumo alto de alcohol en fin de semana (Walc >= 4)[span_4](start_span)[span_4](end_span)
estudiantes_walc_alto = df[df['Walc'] >= 4]
print(f"Estudiantes con Walc >= 4: {len(estudiantes_walc_alto)}")

# 3.D Selecciona a los estudiantes con studytime mayor que la mediana[span_5](start_span)[span_5](end_span)
mediana_studytime = df['studytime'].median()
estudiantes_mucho_estudio = df[df['studytime'] > mediana_studytime]
print(f"Estudiantes que estudian más de la mediana ({mediana_studytime}): {len(estudiantes_mucho_estudio)}\n")


# --- 4. Creación de columnas nuevas ---

# 4.A Crea columna high_alcohol (Verdadera si Dalc + Walc >= 6, falsa si no)[span_6](start_span)[span_6](end_span)
df['high_alcohol'] = (df['Dalc'] + df['Walc']) >= 6

# 4.B Crea columna categórica alcohol_category[span_7](start_span)[span_7](end_span)
# Para esto, definimos las condiciones y usamos 'loc' para asignar el texto correspondiente
suma_alcohol = df['Dalc'] + df['Walc']

df.loc[suma_alcohol <= 3, 'alcohol_category'] = 'Low'
df.loc[(suma_alcohol >= 4) & (suma_alcohol <= 5), 'alcohol_category'] = 'Moderate'
df.loc[suma_alcohol >= 6, 'alcohol_category'] = 'High'

print("✅ Secciones 3 y 4 completadas con éxito.")
print("Aquí tienes una muestra de las nuevas columnas creadas:")
df[['student_id', 'avg_grade', 'high_alcohol', 'alcohol_category']].head(10)
```

```python
# --- 5. Agrupaciones ---

# 5.A Calcula el promedio de avg_grade por sexo (sex)[span_1](start_span)[span_1](end_span)
promedio_calif_sexo = df.groupby('sex')['avg_grade'].mean()
print("--- 5.A Promedio de calificación por sexo ---")
print(promedio_calif_sexo, "\n")

# 5.B Calcula la mediana de Walc por sexo[span_2](start_span)[span_2](end_span)
mediana_walc_sexo = df.groupby('sex')['Walc'].median()
print("--- 5.B Mediana de consumo de alcohol en fin de semana (Walc) por sexo ---")
print(mediana_walc_sexo, "\n")

# 5.C Calcula el número de estudiantes por categoría de alcohol (alcohol_category)[span_3](start_span)[span_3](end_span)
# Usamos value_counts() que es ideal para contar categorías
conteo_alcohol = df['alcohol_category'].value_counts()
print("--- 5.C Estudiantes por categoría de alcohol ---")
print(conteo_alcohol, "\n")

# 5.D Calcula el promedio del tiempo de estudio (studytime) por categoría de alcohol[span_4](start_span)[span_4](end_span)
estudio_por_alcohol = df.groupby('alcohol_category')['studytime'].mean()
print("--- 5.D Promedio de tiempo de estudio por categoría de alcohol ---")
print(estudio_por_alcohol, "\n")


# --- 6. Ordenamiento y top-N ---

# 6.A Ordena a los estudiantes por avg_grade de mayor a menor[span_5](start_span)[span_5](end_span)
df_ordenado = df.sort_values(by='avg_grade', ascending=False)

# 6.B Selecciona los 10 estudiantes con mayor avg_grade y guárdalos en top10[span_6](start_span)[span_6](end_span)
# Usamos .copy() para evitar advertencias de Pandas más adelante
top10 = df_ordenado.head(10).copy()

# 6.C Selecciona los 10 estudiantes con menor avg_grade y guárdalos en bottom10[span_7](start_span)[span_7](end_span)
bottom10 = df_ordenado.tail(10).copy()

print("✅ Secciones 5 y 6 completadas con éxito.")
print(f"Se crearon las tablas 'top10' con {len(top10)} filas y 'bottom10' con {len(bottom10)} filas.")
```

```python
# --- 7. Integración de resultados (concat y merge) ---

# 7.A Concatenación[span_3](start_span)[span_3](end_span)
# Agregamos la columna 'grupo' a cada tabla[span_4](start_span)[span_4](end_span)
top10['grupo'] = 'top'
bottom10['grupo'] = 'bottom'
# Unimos una debajo de la otra[span_5](start_span)[span_5](end_span)
ranking20 = pd.concat([top10, bottom10])

# 7.B Tabla resumen[span_6](start_span)[span_6](end_span)
# Calculamos promedios de avg_grade, Walc y Dalc por sexo[span_7](start_span)[span_7](end_span)
resumen_sex = df.groupby('sex')[['avg_grade', 'Walc', 'Dalc']].mean().reset_index()

# 7.C Unión (merge)[span_8](start_span)[span_8](end_span)
# Hacemos el merge tipo 'left' usando 'sex' como llave[span_9](start_span)[span_9](end_span)
ranking20_enriquecido = pd.merge(ranking20, resumen_sex, on='sex', how='left', suffixes=('', '_promedio_general'))

# 7.D Indica cuántas filas y columnas tienen las nuevas tablas[span_10](start_span)[span_10](end_span)
print("--- 7.D Dimensiones de las tablas integradas ---")
print(f"ranking20: {ranking20.shape[0]} filas y {ranking20.shape[1]} columnas.")
print(f"ranking20_enriquecido: {ranking20_enriquecido.shape[0]} filas y {ranking20_enriquecido.shape[1]} columnas.\n")


# --- 8. Cruce de variables ---

# 8.A Tabla cruzada entre sex y alcohol_category[span_11](start_span)[span_11](end_span)
cruce_sex_alcohol = pd.crosstab(df['sex'], df['alcohol_category'])
print("--- 8.A Tabla cruzada: Sexo vs Categoría de Alcohol ---")
print(cruce_sex_alcohol, "\n")

# 8.B Tabla cruzada entre alcohol_category y studytime[span_12](start_span)[span_12](end_span)
cruce_alcohol_estudio = pd.crosstab(df['alcohol_category'], df['studytime'])
print("--- 8.B Tabla cruzada: Categoría de Alcohol vs Tiempo de Estudio ---")
print(cruce_alcohol_estudio, "\n")


# --- 9. Exportación ---

# Guardamos los 4 DataFrames en archivos CSV (ponemos index=False para que no guarde los números de fila como una columna extra)[span_13](start_span)[span_13](end_span)
top10.to_csv('top10.csv', index=False)
bottom10.to_csv('bottom10.csv', index=False)
ranking20.to_csv('ranking20.csv', index=False)
ranking20_enriquecido.to_csv('ranking20_enriquecido.csv', index=False)

print("✅ ¡Misión cumplida! Secciones 7, 8 y 9 terminadas.")
print("Revisa la barra lateral izquierda de Visual Studio Code: ahí verás tus 4 archivos CSV nuevos listos para entregar.")
```
